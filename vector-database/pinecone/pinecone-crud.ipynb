```python
from pinecone import Pinecone
import os

pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))

idx = pc.Index("insert")

vectors = [
    ("A", [0.1, 0.1, 0.1]),
    ("B", [0.2, 0.2, 0.2]),
    ("C", [0.3, 0.3, 0.3]),
    ("D", [0.4, 0.4, 0.4]),
    ("E", [0.5, 0.5, 0.5]),
]

idx.upsert(vectors=vectors)
```

```python
idx.update(id="A", values=[0.99, 0.99, 0.99])
```

```python
idx.query(vector=[0.2, 0.5, 0.2], top_k=2, include_values=True)
```

```python
idx.fetch(ids=["A"])
```

```python
idx.delete(ids=["vec1"])

# Delete all vectors
idx.delete(delete_all=True)
```

```python
from pinecone import Pinecone, ServerlessSpec
import os

pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))

pc.create_index(
    name="insert",
    dimension=3,
    spec=ServerlessSpec(
        cloud="aws",
        region="us-east-1",
    )
)

pc.list_indexes()
```

```python
# The first element of the tuple is the id, the second is the vector
data = [("vec1", [1.0, 2.0, 3.0]), ("vec2", [11.0, 12.0, 13.0]), ("vec3", [31.0, 32.0, 33.0])]
```

## Insert/update data

```python
idx = pc.Index("insert")

idx.upsert(data)
```

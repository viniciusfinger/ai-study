```python
from pinecone import Pinecone
import os

pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))

idx = pc.Index("test")

idx.upsert([
    ("1", [0.1, 0.2, 0.3], {"topic": "subject", "year": 2020}),
    ("2", [0.4, 0.5, 0.6], {"topic": "other", "year": 2021}),
    ("3", [0.7, 0.8, 0.9], {"topic": "body", "year": 2022}),
    ("4", [0.2, 0.3, 0.4], {"topic": "body"}),
    ("5", [0.5, 0.6, 0.7], {"topic": "subject"}),
])
```

```python
# Querying

idx.query(
    vector=[0, 0, 0], 
    top_k=2,
    include_values=True,
    include_metadata=True, 
    filter= {
        "topic": {"$eq": "subject"}
    }
    )
```

```python
idx.query(
    vector=[0, 0, 0], 
    top_k=2,
    include_values=True,
    include_metadata=True, 
    filter= {
        "year": {"$gt": 2021}
    }
    )
```

```python
idx.update(id="1", set_metadata={"topic": "body"})
```

```python
idx.query(id="1", top_k=2, include_metadata=True)
```

```python
idx.delete(filter = {
    "topic": {"$eq": "body"}
})
```

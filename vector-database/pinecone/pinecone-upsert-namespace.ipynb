```python
from pinecone import Pinecone
import os
import numpy as np

pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))

idx = pc.Index("test")
```

```python
# Create dummy data for example
dimensions = 3
emails_with_subject = 20
email_with_body = 45
emails_with_other = 45

subject_vectors = np.random.default_rng().random((emails_with_subject, dimensions)).tolist()
body_vectors = np.random.default_rng().random((email_with_body, dimensions)).tolist()
other_vectors = np.random.default_rng().random((emails_with_other, dimensions)).tolist()

subject_ids = [f"subject-{i}" for i in range(emails_with_subject)]
body_ids = [f"body-{i}" for i in range(email_with_body)]
other_ids = [f"other-{i}" for i in range(emails_with_other)]

subject_vectors = list(zip(subject_ids, subject_vectors))
body_vectors = list(zip(body_ids, body_vectors))
other_vectors = list(zip(other_ids, other_vectors))

subject_vectors[:5]
```

```python
idx.upsert(vectors=subject_vectors, namespace="subject")

idx.upsert(vectors=body_vectors, namespace="body")

idx.upsert(vectors=other_vectors, namespace="other")

```
